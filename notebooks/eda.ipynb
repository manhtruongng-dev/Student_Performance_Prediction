```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Ghi dữ liệu và hiển thị các dòng đầu tiên
df = pd.read_csv(r"../data/student-mat.csv",sep=";")
df.head(10)
```

```python
# Thống kê cơ bản 
df.describe()
```

```python
#Kiểm tra dữ liệu NaN hoặc None
df.isna().sum()
```

```python
# Kiểm tra số lượng trùng lặp dữ liệu
df.duplicated().sum()
```

```python
# Biểu đồ phân tích điểm cuối kì (G3)
# plt.figure(figsize=(6,3))
# sns.histplot(df['G3'],kde = True)
# plt.title("Phân tích điểm cuối kì (G3)", fontsize=13)
# plt.xlabel("Điểm số")
# plt.ylabel("Số lượng")
# plt.show()

fig,axes = plt.subplots(1,2,figsize =(15,6))
sns.histplot(df['G3'],kde = True,ax=axes[0],bins=20)
axes[0].bar_label(axes[0].containers[0], fmt='%d')
axes[0].set_title("Biểu đồ cột")
axes[0].set_xlabel("Điểm số")
axes[0].set_ylabel("Số lượng")
sns.boxplot(df['G3'],ax=axes[1])
axes[1].set_ylabel("Điểm số")
axes[1].set_title("Biểu đồ hộp")

fig.suptitle("Phân tích điểm cuối kì (G3)",fontsize=15)
fig.tight_layout()
plt.show()
```

Nhận xét Biểu đồ 1:  
    &emsp; - Biểu đồ bên trái cho thấy có rất nhiều học sinh có điểm từ 8-15, cà có 38 học sinh đạt điểm 0 cuối kì  
    &emsp; - Biểu đồ bên phải:  
        &emsp;&emsp; + Vạch ngang ngay tại điểm 11 và chia hộp thành 2 phần bằng nhau cho thấy điểm từ 11 điểm trở xuống bằng điểm từ 11 điểm trở lên. Dựa vào biểu đồ ta nhận xét mức học lực trung bình của trường nằm ở mức trung bình khá  
        &emsp;&emsp; + Đỉnh hộp tại điểm 14, đáy hộp tại điểm 8 cho thấy 50% học sinh có điểm sàn tập trung từ 8 đến 14 

```python
# Biểu đồ 2: Các biểu đồ hộp các đặc trưng số
num_cols = ['age','absences','G1','G2','G3']
fig, axes = plt.subplots(1, len(num_cols), figsize=(16, 5))

for i, col in enumerate(num_cols):
    sns.boxplot(data=df, y=col, ax=axes[i], color='lightseagreen')
    axes[i].set_title(f"Phân bố {col}")
    axes[i].set_ylabel(col)

plt.suptitle("Biểu đồ hộp các đặc trưng số", fontsize=16)
plt.tight_layout()
plt.show()
```

Nhận xét biểu đồ 2:  
  &emsp;- Ta nhìn thấy các chấm tròn tại các đặc trưng age, absences, G2, các chấm này cho biết các ngoại lai  
    &emsp;&emsp;+ Đặc trưng absence nhiều chấm tròn, có thể thấy có độ nhiễu rất nghiêm trọng, dải chứa lượng lớn ngoại lai từ 20 đến 75 buổi.  
    &emsp;&emsp;+ Đặc trưng age ta thấy có 1 chấm ngoại lai 22 tuổi - tuổi này với học sinh phổ thông là hơi hiếm nhưng có thể đó là bảo lưu hoặc học lại rất cao do số lượng chỉ có 1, nên ta vẫn giữ đặc trưng này  
    &emsp;&emsp;+ Đặc trưng G2 ta thấy có 1 chấm ngoại lai 0 điểm - điểm này không phản ảnh đúng năng lực học tập, có thể là do bỏ thi hoặc vi phạm quy chế thi.  
  &emsp;- Còn lại 2 đặc trưng G1, G3 có bộ điểm phân bố đồng đều và không có ngoại lai.

```python
# Biểu đồ 3: Các biểu đồ cột các đặc trưng chữ
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

sns.countplot(data=df, x='sex', hue='sex', ax=axes[0,0], palette='Set2', legend=False)
axes[0,0].set_xlabel("Giới tính (sex)")
axes[0,0].set_ylabel("Số lượng")
axes[0,0].set_title("Phân bố Giới tính (F: nữ, M: nam)")

sns.countplot(data=df, x='address', hue='address', ax=axes[0,1], palette='Set2', legend=False)
axes[0,1].set_xlabel("Nơi sống (address)")
axes[0,1].set_ylabel("Số lượng")
axes[0,1].set_title("Phân bố Nơi sống (U: Thành thị, R: Nông thôn)")

sns.countplot(data=df, x='schoolsup', hue='schoolsup', ax=axes[1,0], palette='Set2', legend=False)
axes[1,0].set_xlabel("Hỗ trợ từ trường (schoolsup)")
axes[1,0].set_ylabel("Số lượng")
axes[1,0].set_title("Tình trạng nhận hỗ trợ từ trường")

sns.countplot(data=df, x='higher', hue='higher', ax=axes[1,1], palette='Set2', legend=False)
axes[1,1].set_xlabel("Nguyện vọng Đại học")
axes[1,1].set_ylabel("Số lượng")
axes[1,1].set_title("Nguyện vọng học lên Đại học")

plt.suptitle("Biểu đồ cột các đặc trưng chữ", fontsize=16)
plt.tight_layout()
plt.show()
```

Nhận xét biểu đồ 3:  
     &emsp;- Ta thấy ngoại trừ biểu đồ giới tính ra thì các biểu đồ còn lại đều mất cân bằng dữ liệu

```python
#Biểu đồ 3: Ma trận tương quan
plt.figure(figsize=(13,7))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True,cmap="coolwarm")
plt.title("Biểu đồ tương quan",fontsize=15)

plt.show()
```

Nhận xét Biểu đồ 3: (dấu '+' tương quan thuận, dấu '-' tương quan nghịch)  
    &emsp;- Màu đỏ G1, G2 cho thấy ảnh hưởng mạnh tới G3  
    &emsp;- Các giá trị gần 0 như famrel, freetime,Dalc,Walc, ít ảnh hưởng tới G3 do nó gần tiệm cận với giá trị 0, và studytime(0.098) tương quan tương đối thấp với G3  
    &emsp;- Các giá trị ảnh hưởng khác  
        &emsp;&emsp;+ age(-0.16) cho thấy khả năng học sinh bảo lưu hoặc học sinh học lại  
        &emsp;&emsp;+ Medu(0.22) cho thấy học vấn mẹ càng cao thì điểm G3 tăng nhiều  
        &emsp;&emsp;+ Fedu(0.15) cho thấy học vấn cha càng cao thì điểm G3 tăng nhiều  
        &emsp;&emsp;+ Failures(-0.36) cho thấy càng rớt nhiều thì điểm G3 sẽ giảm nhiều  



```python
#Biểu đồ 4: Các biểu đồ phân tích mối quan hệ đặc trưng với nhãn
fig, axes = plt.subplots(2, 3, figsize=(18, 12))

sns.regplot(data=df, x='G1', y='G3', ax=axes[0,0], color='royalblue', scatter_kws={'alpha':0.6})
axes[0,0].set_title("Phân tán: Điểm G1 và G3")
axes[0,0].set_xlabel("Điểm kì 1 (G1)")
axes[0,0].set_ylabel("Điểm cuối kì (G3)")

sns.regplot(data=df, x='G2', y='G3', ax=axes[1,0], color='crimson', scatter_kws={'alpha':0.6})
axes[1,0].set_title("Phân tán: Điểm G2 và G3")
axes[1,0].set_xlabel("Điểm kì 2 (G2)")
axes[1,0].set_ylabel("Điểm cuối kì (G3)")

sns.regplot(data=df, x='age', y='G3', ax=axes[0,2], color='purple', x_jitter=0.15, scatter_kws={'alpha':0.5})
axes[0,2].set_title("Phân tán: Tuổi (age) và G3")
axes[0,2].set_xlabel("Tuổi (age)")
axes[0,2].set_ylabel("Điểm cuối kì (G3)")

sns.regplot(data=df, x='Medu', y='G3', ax=axes[0,1], color='seagreen', x_jitter=0.15, scatter_kws={'alpha':0.5})
axes[0,1].set_title("Phân tán: Học vấn mẹ (Medu) và G3")
axes[0,1].set_xlabel("Học vấn mẹ (Medu)")
axes[0,1].set_ylabel("Điểm cuối kì (G3)")

sns.regplot(data=df, x='Fedu', y='G3', ax=axes[1,1], color='darkorange', x_jitter=0.15, scatter_kws={'alpha':0.5})
axes[1,1].set_title("Phân tán: Học vấn cha (Fedu) và G3")
axes[1,1].set_xlabel("Học vấn cha (Fedu)")
axes[1,1].set_ylabel("Điểm cuối kì (G3)")

sns.regplot(data=df, x='failures', y='G3', ax=axes[1,2], color='saddlebrown', x_jitter=0.1, scatter_kws={'alpha':0.5})
axes[1,2].set_title("Phân tán: Số lần trượt (failures) và G3")
axes[1,2].set_xlabel("Số lần trượt (failures)")
axes[1,2].set_ylabel("Điểm cuối kì (G3)")
plt.suptitle("Các biểu đồ phân tích mối quan hệ đặc trưng với nhãn", fontsize=16)
plt.tight_layout()
plt.show()

```

Nhận xét biểu đồ 4:  
    &emsp;- Ta thấy 2 biểu đồ bên trái có tập các dữ liệu trải đều trên đường thẳng, khẳng định điểm G1, G2 là các nhân tố cốt lõi ảnh hưởng mạnh tới điểm G3  
    &emsp;- Ta thấy 4 biểu đồ bên phải có tập dữ liệu phân bố rời rạc thành cụm dọc và phân tán xa đường thẳng nhưng đường thẳng vẫn có độ dốc, cho thấy các nhân đố ấy có tác động tới điểm G3 nhưng không mang quyết định tuyệt đối, như điểm G1 và G2 cao học sinh vẫn có điểm G3 cao.
