```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Ghi dữ liệu và hiển thị các dòng đầu tiên
df = pd.read_csv(r"../data/student-mat.csv",sep=";")
df.head(10)
```

```python
# Thống kê cơ bản 
df.describe()
```

```python
#Kiểm tra dữ liệu NaN hoặc None
df.isna().sum()
```

```python
# Kiểm tra số lượng trùng lặp dữ liệu
df.duplicated().sum()
```

```python
# Biểu đồ phân tích điểm cuối kì (G3)
# plt.figure(figsize=(6,3))
# sns.histplot(df['G3'],kde = True)
# plt.title("Phân tích điểm cuối kì (G3)", fontsize=13)
# plt.xlabel("Điểm số")
# plt.ylabel("Số lượng")
# plt.show()

fig,axes = plt.subplots(1,2,figsize =(15,4))
sns.histplot(df['G3'],kde = True,ax=axes[0],bins=20)
axes[0].bar_label(axes[0].containers[0], fmt='%d')
axes[0].set_title("Biểu đồ cột")
axes[0].set_xlabel("Điểm số")
axes[0].set_ylabel("Số lượng")
sns.boxplot(df['G3'],ax=axes[1])
axes[1].set_ylabel("Điểm số")
axes[1].set_title("Biểu đồ hộp")

fig.suptitle("Phân tích điểm cuối kì (G3)",fontsize=15)
fig.tight_layout()
plt.show()
```

Nhận xét Biểu đồ 1: 
    - Biểu đồ bên trái cho thấy có rất nhiều học sinh có điểm từ 8-15, cà có 38 học sinh đạt điểm 0 cuối kì
    - Biểu đồ bên phải:
        + Vạch ngang ngay tại điểm 11 và chia hộp thành 2 phần bằng nhau cho thấy điểm từ 11 điểm trở xuống bằng điểm từ 11 điểm trở lên. Dựa vào biểu đồ ta nhận xét mức học lực trung bình của trường nằm ở mức trung bình khá
        + Đỉnh hộp tại điểm 14, đáy hộp tại điểm 8 cho thấy 50% học sinh có điểm sàn tập trung từ 8 đến 14 

```python
#Biểu đồ 2: Ma trận tương quan
plt.figure(figsize=(13,7))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True,cmap="coolwarm")
plt.title("Biểu đồ tương quan",fontsize=15)

plt.show()
```
