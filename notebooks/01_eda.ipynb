```python
#Số lượng giá trị phân biệt đặc trưng phân loại
df.select_dtypes(include=['object','category','string']).nunique()
```

Nhận xét:  
    - Có Mjob, Fjob, reason, guardian >2 cho thấy đây là các cột đặc trưng phân loại thuộc nhóm đa lớp

```python
cat_cols = df.select_dtypes(include=['object', 'category', 'string']).columns
binary_cols = [col for col in cat_cols if df[col].nunique() == 2]

fig, axes = plt.subplots(3, 4, figsize=(16, 12))
axes = axes.flatten()

for i, col in enumerate(binary_cols[:12]):
    sns.countplot(x=col, data=df, ax=axes[i], palette='viridis', hue=col, legend=False)
    axes[i].set_title(col)
fig.suptitle("Các biểu đồ cột phân tích sự cân bằng dữ liệu của đặc trưng phân loại nhóm nhị phân",fontsize=15)
plt.tight_layout()
plt.show()
```

Nhận xét: 
 - Trong 13 đặc trưng phân loại thuộc nhóm nhị phân, biểu đồ trên vẽ 12 trên 13 đặc trưng, nhưng trong số đó sự cân bằng dữ liệu nó xuất hiện 1 tới 2 biểu đồ, cho thấy dữ liệu ở các đặc trưng phân loại thuộc nhóm nhị phân không cân bằng  
 - Trong bước lựa chọn các đặc trưng sẽ bỏ các đặc trưng phân loại

```python
#Biểu đồ 4: Ma trận tương quan
plt.figure(figsize=(13,7))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True,cmap="coolwarm")
plt.title("Biểu đồ tương quan",fontsize=15)

plt.show()
```

Nhận xét Biểu đồ 4: (dấu '+' tương quan thuận, dấu '-' tương quan nghịch)  
    &emsp;- Màu đỏ G1, G2 cho thấy ảnh hưởng mạnh tới G3  
    &emsp;- Các giá trị gần 0 như famrel, freetime,Dalc,Walc, ít ảnh hưởng tới G3 do nó gần tiệm cận với giá trị 0, và studytime(0.098) tương quan tương đối thấp với G3  
    &emsp;- Các giá trị ảnh hưởng khác  
        &emsp;&emsp;+ age(-0.16) cho thấy khả năng học sinh bảo lưu hoặc học sinh học lại  
        &emsp;&emsp;+ Medu(0.22) cho thấy học vấn mẹ càng cao thì điểm G3 tăng nhiều  
        &emsp;&emsp;+ Fedu(0.15) cho thấy học vấn cha càng cao thì điểm G3 tăng nhiều  
        &emsp;&emsp;+ Failures(-0.36) cho thấy càng rớt nhiều thì điểm G3 sẽ giảm nhiều  
Kết luận:
 - Nhóm "Cột trụ" (Tương quan rất mạnh): {G1, G2}. Đây là các biến không thể thiếu vì tương quan với G3 lên đến 0.8-0.9.
 - Nhóm "Cảnh báo" (Tương quan nghịch): {failures, age, traveltime, goout}. Những biến này cho thấy xu hướng "càng tăng thì điểm càng giảm". Đây là bằng chứng quan trọng cho thấy mô hình hiểu được quy luật tiêu cực.
 - Nhóm "Hỗ trợ" (Tương quan nhẹ): {Medu, Fedu, studytime}. Dù tương quan nhỏ nhưng vẫn > 0. Việc giữ lại chúng làm mô hình đầy đủ thông tin hơn.



```python
fig, axes = plt.subplots(3, 3, figsize=(18, 18))
sns.regplot(data=df, x='G1', y='G3', ax=axes[0,0], color='royalblue', scatter_kws={'alpha':0.6})
axes[0,0].set_title("Phân tán: Điểm G1 và G3")

sns.regplot(data=df, x='Medu', y='G3', ax=axes[0,1], color='seagreen', x_jitter=0.15, scatter_kws={'alpha':0.5})
axes[0,1].set_title("Phân tán: Học vấn mẹ (Medu) và G3")

sns.regplot(data=df, x='age', y='G3', ax=axes[0,2], color='purple', x_jitter=0.15, scatter_kws={'alpha':0.5})
axes[0,2].set_title("Phân tán: Tuổi (age) và G3")

# Hàng 2: G2, Fedu, failures
sns.regplot(data=df, x='G2', y='G3', ax=axes[1,0], color='crimson', scatter_kws={'alpha':0.6})
axes[1,0].set_title("Phân tán: Điểm G2 và G3")

sns.regplot(data=df, x='Fedu', y='G3', ax=axes[1,1], color='darkorange', x_jitter=0.15, scatter_kws={'alpha':0.5})
axes[1,1].set_title("Phân tán: Học vấn cha (Fedu) và G3")

sns.regplot(data=df, x='failures', y='G3', ax=axes[1,2], color='saddlebrown', x_jitter=0.1, scatter_kws={'alpha':0.5})
axes[1,2].set_title("Phân tán: Số lần trượt (failures) và G3")

# Hàng 3: traveltime, goout, studytime (Các cột mới thêm vào)
sns.regplot(data=df, x='traveltime', y='G3', ax=axes[2,0], color='teal', x_jitter=0.1, scatter_kws={'alpha':0.5})
axes[2,0].set_title("Phân tán: Thời gian đi lại và G3")

sns.regplot(data=df, x='goout', y='G3', ax=axes[2,1], color='magenta', x_jitter=0.1, scatter_kws={'alpha':0.5})
axes[2,1].set_title("Phân tán: Đi chơi (goout) và G3")

sns.regplot(data=df, x='studytime', y='G3', ax=axes[2,2], color='navy', x_jitter=0.1, scatter_kws={'alpha':0.5})
axes[2,2].set_title("Phân tán: Thời gian học (studytime) và G3")

plt.suptitle("Các biểu đồ phân tích mối quan hệ đặc trưng với nhãn", fontsize=20)
plt.tight_layout(rect=[0, 0.03, 1, 0.97])
plt.show()
```

Nhận xét biểu đồ 5:  
    &emsp;- Nhóm nhân tố cốt lõi (G1, G2): Các biểu đồ cho thấy dữ liệu tập trung dày đặc, bám sát đường xu hướng tuyến tính với độ dốc cao. Điều này khẳng định điểm quá trình (G1, G2) là những nhân tố quyết định và có khả năng dự báo mạnh mẽ nhất đối với điểm cuối kỳ (G3).  
    &emsp;- Nhóm nhân tố phụ trợ (Medu, Fedu, studytime): Các biểu đồ này ghi nhận độ phân tán dữ liệu lớn hơn và dữ liệu phân bố thành cụm dọc theo các giá trị nguyên. Tuy nhiên, đường xu hướng vẫn thể hiện độ dốc dương, cho thấy học vấn của cha mẹ và thời gian tự học có tác động tích cực nhất định, dù không mang tính quyết định tuyệt đối.  
    &emsp;- Nhóm nhân tố cảnh báo (failures, age, goout, traveltime): Các biểu đồ ghi nhận xu hướng dốc âm, cho thấy mối quan hệ tỷ lệ nghịch với điểm số. Đặc biệt, các biến như failures (số lần trượt) và goout (đi chơi) thể hiện rõ xu hướng điểm G3 giảm dần khi giá trị đặc trưng tăng. Mặc dù dữ liệu phân tán rời rạc, những biểu đồ này cung cấp tín hiệu quan trọng giúp nhận đối tượng học tập kém.
